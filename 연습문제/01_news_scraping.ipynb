```python
import requests
from bs4 import BeautifulSoup


url = "https://news.daum.net/economy"

print(url)

# 요청 헤더 설정 : 브라우저 정보
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}

# requests 의 get() 함수 호출하기

res = requests.get(url, headers=req_header)
print(res.status_code, res.ok)

# 응답(response)이 OK 이면

if res.ok:
# response를 utf-8로 encoding
    res.encoding = 'utf-8'

# 응답 (response)에서 text 추출
    html = res.text

# BeautifulSoup 객체 생성
    soup = BeautifulSoup(html, 'html.parser')

# CSS 선택자
    li_tags = soup.select('ul.list_newsheadline2 li')

    print(type(li_tags), len(li_tags))


    a_tags = soup.select("ul.list_newsheadline2 li a")

    for a_tag in a_tags:
        title = a_tag.select_one("strong.tit_txt").text.strip()
        link = a_tag['href']
        print(link)
        print(title)
# 응답(response)이 Error 이면 status code 출력

else:
    print(f'Error Code = {res.status_code}')    
```

```python
import requests
from bs4 import BeautifulSoup

section_dict = {
    '기후/환경': 'climate',
    '사회': 'society',
    '경제': 'economy',
    '정치': 'politics',
    '국제': 'world',
    '문화': 'culture',
    '생활': 'life',
    'IT/과학': 'tech',
    '인물': 'people'
}

# 함수 선언
def print_news(section_name):

    # section 이름으로 영문 문자열 추출
    section = section_dict[section_name]

    # URL 생성
    url = f"https://news.daum.net/{section}"

    print(f"======> {url} {section_name} 뉴스 <======")

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
    }

    # requests 의 get() 함수 호출하기
    res = requests.get(url, headers=req_header)

    # 응답(response)이 OK 이면
    if res.ok:

        # response를 utf-8로 encoding
        res.encoding = 'utf-8'

        # 응답(response)에서 text 추출
        html = res.text

        # BeautifulSoup 객체 생성
        soup = BeautifulSoup(html, 'html.parser')

        # CSS 선택자
        a_tags = soup.select("ul.list_newsheadline2 li a")

        # <a> 태그 리스트 순회하기
        for a_tag in a_tags:

            strong_tag = a_tag.select_one("strong.tit_txt")

            if strong_tag:
                title = strong_tag.text.strip()
                link = a_tag['href']

                print(link)
                print(title)

    # 응답(response)이 Error 이면 status code 출력
    else:
        print(f'Error Code = {res.status_code}')

#test
print_news('경제')
    
```
