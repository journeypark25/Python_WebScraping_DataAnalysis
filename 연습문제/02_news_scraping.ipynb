뉴스기사의 Image를 출력 하세요 
1) Image의 도메인이름이 포함된 url과 src 속성의 img 경로를 합치려면 urljoin 함수를 사용하세요.

3) img 엘리먼트의 존재 여부를 체크하신 후에 src 속성의 이미지를 경로를 추출하기
  => Image 가 없는 뉴스도 있기 때문에 


```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from IPython.display import Image, display

section_dict = {
    '최신뉴스': 'n0100',
    '정치': 'n0101',
    '경제': 'n0102',
    '사회': 'n0103',
    '세계': 'n0104',
    'IT/과학': 'n0105'
}

# 함수 선언
def print_nate_news(section_name):

    # section 이름으로 mid 값 추출
    mid = section_dict[section_name]

    # URL 생성
    url = f"https://news.nate.com/recent?mid={mid}"

    print(f"======> {url} {section_name} 뉴스 <======")

    # 요청 헤더 설정 : 브라우저 정보
    req_header = {
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
    }

    # requests 의 get() 함수 호출하기
    res = requests.get(url, headers=req_header)

    # 응답(response)이 OK 이면
    if res.ok:

        # 응답 페이지의 인코딩 자동 설정
        res.encoding = res.apparent_encoding

        # 응답(response)에서 text 추출
        html = res.text

        # BeautifulSoup 객체 생성
        soup = BeautifulSoup(html, 'html.parser')

        # 뉴스 기사 목록
        div_tags = soup.select("div.mduSubjectList")

        print(type(div_tags), len(div_tags))

        # 뉴스 기사 리스트 중 5개만 순회하기
        for div_tag in div_tags[:5]:

            # <a> 태그 추출
            a_tag = div_tag.find('a')

            if a_tag:
                # 기사 제목
                title = a_tag.text.strip()

                # 기사 링크
                link = urljoin(url, a_tag['href'])

                # img 태그 추출
                img_tag = div_tag.find('img')

                # 이미지가 있는 경우
                if img_tag:
                    src = img_tag['src']

                    # 이미지 URL 완성
                    img_url = urljoin(url, src)

                    # 이미지 출력
                    display(Image(url=img_url))

                # 제목, 링크 출력
                print(title)
                print(link)
                print()

    # 응답(response)이 Error 이면 status code 출력
    else:
        print(f'Error Code = {res.status_code}')


# 모든 섹션 출력
for section_name in section_dict:
    print_nate_news(section_name)
```

* 2-2 하나의 네이버 웹툰과 1개의 회차에 대한 Image 다운로드 하기 (필수)
:  하나의 웹툰의 제목(title)과 회차번호(no),회차의URL(url) 을 입력으로 받는 함수를 선언합니다. 
   def download_one_episode(title,no,url):

아래와 같이 호출합니다.
download_one_episode('일렉시드',341,'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed')

img\일렉시드\341 디렉토리가 생성되며 , 
그 디렉토리 아래에 웹툰 image들이 다운로드 되도록 해주세요.


```python
import requests
from bs4 import BeautifulSoup
import os


# 함수 선언
def download_one_episode(title, no, url):

    # 요청 헤더 설정
    req_header = {
        'referer': url,
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36'
    }

    # 이미지 저장 폴더
    imgdir_name = os.path.join('img', title, str(no))

    # 이미지 저장 폴더가 없으면 생성
    # sub 디렉토리도 생성
    os.makedirs(imgdir_name, exist_ok=True)

    # 웹 페이지 요청 및 확인
    res = requests.get(url, headers=req_header)

    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    # 이미지 URL 추출
    soup = BeautifulSoup(res.text, 'html.parser')

    img_url_list = [
        img_tag['src']
        for img_tag in soup.select("img[src*='IMAG01']")
    ]

    print(f'이미지 개수 = {len(img_url_list)}')

    # 이미지 다운로드
    for img_url in img_url_list:

        res = requests.get(img_url, headers=req_header)

        if res.ok:
            img_data = res.content

            file_path = os.path.join(
                imgdir_name,
                os.path.basename(img_url)
            )

            with open(file_path, 'wb') as file:
                print(
                    f'Writing to {file_path} '
                    f'({len(img_data):,} bytes)'
                )

                file.write(img_data)

        else:
            print(
                f'Error Code = {res.status_code} '
                f'for {img_url}'
            )

# 호출
download_one_episode(
    '일렉시드',
    341,
    'https://comic.naver.com/webtoon/detail?titleId=717481&no=341&week=wed'
)
```
